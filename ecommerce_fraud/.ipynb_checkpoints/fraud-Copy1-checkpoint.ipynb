```python
from sklearn.metrics import accuracy_score
from sklearn.metrics import precision_score
from sklearn.metrics import recall_score
from sklearn.metrics import f1_score
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report
```

```python
accuracy=[]
precision=[]
recall=[]
fscore=[]
```

```python
def calculateMetrics(algorithm, predict, testY):
    import numpy as np
    import seaborn as sns
    import matplotlib.pyplot as plt
    from sklearn.metrics import (
        precision_score, recall_score, f1_score,
        accuracy_score, classification_report, confusion_matrix
    )

    # Ensure integer arrays
    predict = np.array(predict).flatten().astype(int)
    testY = np.array(testY).flatten().astype(int)

    # ---- UPDATE: 7-class labels ----
    
    labels_name = [
            'account_takeover', 'identity_theft', 'synthetic_account',
       'promotion_abuse', 'legit', 'refund_fraud', 'payment_fraud'
    ]

    labels_int = list(range(len(labels_name)))  # [0,1,2,3,4,5,6]

    # Metrics
    p = precision_score(testY, predict, average='macro') * 100
    r = recall_score(testY, predict, average='macro') * 100
    f = f1_score(testY, predict, average='macro') * 100
    a = accuracy_score(testY, predict) * 100

    accuracy.append(a)
    precision.append(p)
    recall.append(r)
    fscore.append(f)

    print(f"{algorithm} Accuracy    : {a:.2f}")
    print(f"{algorithm} Precision   : {p:.2f}")
    print(f"{algorithm} Recall      : {r:.2f}")
    print(f"{algorithm} FSCORE      : {f:.2f}")

    # Classification report for 7 labels
    report = classification_report(
        testY, predict,
        labels=labels_int,
        target_names=labels_name
    )
    print(f"\n{algorithm} classification report\n", report)

    # Confusion matrix
    conf_matrix = confusion_matrix(testY, predict, labels=labels_int)

    plt.figure(figsize=(8, 6))
    ax = sns.heatmap(
        conf_matrix,
        xticklabels=labels_name,
        yticklabels=labels_name,
        annot=True,
        cmap="Blues",
        fmt="g"
    )
    ax.set_ylim([0, len(labels_name)])
    plt.title(f"{algorithm} Confusion Matrix")
    plt.ylabel('True class')
    plt.xlabel('Predicted class')
    plt.tight_layout()
    plt.show()
```

```python
import os
import joblib
from xgboost import XGBClassifier

# Ensure model directory exists
os.makedirs("model", exist_ok=True)

# Define the model file path
model_path = "model/XGBoostClassifier.pkl"

if os.path.exists(model_path):
    # Load pre-trained model
    xgb_model = joblib.load(model_path)
    print("✅ XGBoost model loaded successfully.")
else:
    # Train XGBoost model (x_train, y_train must already be defined)
    xgb_model = XGBClassifier(
        n_estimators=300,
        learning_rate=0.05,
        max_depth=6,
        subsample=0.8,
        colsample_bytree=0.8,
        eval_metric="mlogloss",   # Required for multi-class
        random_state=42
    )
    
    xgb_model.fit(x_train, y_train)

    # Save the trained model
    joblib.dump(xgb_model, model_path)
    print("✅ XGBoost model trained and saved successfully.")

# Predict
y_pred = xgb_model.predict(x_test)

# Evaluate (your custom metrics function)
calculateMetrics("XGBoost", y_pred, y_test)
```

```python
import os
import joblib
from sklearn.ensemble import StackingClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier

# Ensure model directory exists
os.makedirs("model", exist_ok=True)

# Define the model file path
model_path = "model/StackingClassifier.pkl"

# --------------------------
# If model exists → load it
# Else → train & save
# --------------------------
if os.path.exists(model_path):
    stacking_model = joblib.load(model_path)
    print("✅ Stacking Classifier loaded successfully.")
else:
    print("⏳ Training Stacking Classifier...")

    # Base models
    base_models = [
        ("rf", RandomForestClassifier(n_estimators=200, random_state=42)),
        ("dt", DecisionTreeClassifier(random_state=42)),
        ("xgb", XGBClassifier(
            n_estimators=200,
            learning_rate=0.05,
            max_depth=6,
            subsample=0.8,
            colsample_bytree=0.8,
            eval_metric="mlogloss",
            random_state=42
        ))
    ]

    # Meta learner
    meta_model = LogisticRegression(max_iter=500)

    # Stacking model
    stacking_model = StackingClassifier(
        estimators=base_models,
        final_estimator=meta_model,
        stack_method="predict_proba",
        passthrough=False,
        n_jobs=-1
    )

    # Train
    stacking_model.fit(x_train, y_train)

    # Save
    joblib.dump(stacking_model, model_path)
    print("✅ Stacking Classifier trained and saved successfully.")

# --------------------------
# Prediction & Evaluation
# --------------------------
y_pred = stacking_model.predict(x_test)

calculateMetrics("Stacking Classifier", y_pred, y_test)
```
